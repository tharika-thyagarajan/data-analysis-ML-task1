```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Libraries imported successfully!")
```

```python
import pandas as pd

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
display(df.head())
```

```python
print("Number of rows and columns:", df.shape)
print(df.columns.tolist())
display(df.head())
display(df.tail())
```

```python
print("Missing values in each column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("Duplicate rows:", df.duplicated().sum())
```

```python
missing_percent = (df.isnull().mean() * 100).round(2)
display(missing_percent.to_frame("Missing Percentage"))
```

```python
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Convert important numeric columns to numeric values
for col in ["Model Year", "Electric Range", "Base MSRP"]:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```

```python
display(df_clean.isnull().sum().to_frame("Missing After Cleaning"))
```

```python
numeric_cols = df_clean.select_dtypes(include=np.number).columns

stats = df_clean[numeric_cols].describe().T
stats["median"] = df_clean[numeric_cols].median()
stats["variance"] = df_clean[numeric_cols].var()
stats["range"] = (
    df_clean[numeric_cols].max() - df_clean[numeric_cols].min()
)

display(stats)
```

```python
cols = [
    col for col in ["Model Year", "Electric Range", "Base MSRP"]
    if col in df_clean.columns
]

print("Mean:")
display(df_clean[cols].mean())

print("Median:")
display(df_clean[cols].median())

print("Standard deviation:")
display(df_clean[cols].std())

print("Variance:")
display(df_clean[cols].var())

print("Correlation:")
display(df_clean[cols].corr())
```

```python
if "Electric Range" in df_clean.columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(
        data=df_clean,
        x="Electric Range",
        bins=30,
        kde=True
    )
    plt.title("Distribution of Electric Vehicle Range")
    plt.xlabel("Electric Range (miles)")
    plt.ylabel("Number of Vehicles")
    plt.tight_layout()
    plt.show()
```

```python
if "Electric Range" in df_clean.columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df_clean["Electric Range"])
    plt.title("Boxplot of Electric Vehicle Range")
    plt.xlabel("Electric Range (miles)")
    plt.tight_layout()
    plt.show()
```

```python
if "Make" in df_clean.columns:
    top_makes = df_clean["Make"].value_counts().head(10)
    top_makes.sort_values().plot(
        kind="barh", figsize=(8, 5)
    )
    plt.title("Top 10 EV Manufacturers by Record Count")
    plt.xlabel("Number of Records")
    plt.tight_layout()
    plt.show()
```

```python
if {"Electric Vehicle Type", "Electric Range"}.issubset(df_clean.columns):
    plt.figure(figsize=(9, 5))
    sns.boxplot(
        data=df_clean,
        x="Electric Vehicle Type",
        y="Electric Range"
    )
    plt.xticks(rotation=15)
    plt.title("Electric Range by EV Type")
    plt.tight_layout()
    plt.show()
```

```python
if "Model Year" in df_clean.columns:
    df_clean["Model Year"].value_counts().sort_index().plot(
        kind="bar", figsize=(10, 4)
    )
    plt.title("EV Records by Model Year")
    plt.xlabel("Model Year")
    plt.ylabel("Number of Records")
    plt.tight_layout()
    plt.show()
```

```python
cols = [
    c for c in ["Model Year", "Electric Range", "Base MSRP"]
    if c in df_clean.columns
]

plt.figure(figsize=(7, 5))
sns.heatmap(
    df_clean[cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Between Numerical Features")
plt.tight_layout()
plt.show()
```
